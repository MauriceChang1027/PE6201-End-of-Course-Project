# RetinaGuard referral safety evaluation

This notebook runs one fixed referral case for every DR grade. It reports whether OpenRouter output passed the programmatic fact checks or was replaced by the deterministic fallback. No retinal image or identifying patient data is sent to the LLM.

```python
from pathlib import Path
import subprocess
import sys

repo = Path('/content/PE6201-End-of-Course-Project')
if repo.exists():
    subprocess.run(['git', '-C', str(repo), 'pull', '--ff-only'], check=True)
else:
    subprocess.run(['git', 'clone', 'https://github.com/MauriceChang1027/PE6201-End-of-Course-Project', str(repo)], check=True)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-r', str(repo / 'requirements-colab.txt')], check=True)
```

```python
from getpass import getpass
import os

openrouter_key = getpass('OpenRouter API key: ')
run_environment = os.environ.copy()
run_environment['OPENROUTER_API_KEY'] = openrouter_key
results_dir = repo / 'evaluation' / 'llm_results'
subprocess.run([
    sys.executable, '-m', 'scripts.evaluate_referral',
    '--output-dir', str(results_dir),
], cwd=repo, env=run_environment, check=True)
```

```python
import json
import pandas as pd
from IPython.display import display

display(pd.read_csv(results_dir / 'referral_safety_results.csv'))
print(json.dumps(json.loads((results_dir / 'referral_safety_summary.json').read_text()), indent=2))
```
